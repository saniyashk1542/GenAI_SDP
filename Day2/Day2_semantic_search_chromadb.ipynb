```python
from typing import Collection
import chromadb

client=chromadb.PersistentClient(path="./my_vector_db")
collection =client.get_or_create_collection(name="my_docs")

#add docs

collection.add(
    documents=[
        "Python is a programming language used for AI",
        "Machine learning is a subset of artificial intelligence",
        "Cricket World Cup final results 2024 ",
        "Deep Learning uses neural networks",
    ],

    metadatas=[
        {"topic":"tech"},
        {"topic":"ai"},
        {"topic":"sports"},
        {"topic":"ai"},
    ],

    ids=["d1","d2","d3","d4"]
)
```

```python
#Query meaning
results =collection.query(query_texts=["Tell me about AI"],n_results=2)
print("AI Search:",results["documents"])
```

```python
#filter by metadata
filtered =collection.query(
    query_texts=['Tell me about python'],
    where={'topic':'tech'}
)
```

```python
print("Filtered:",filtered["documents"])
```
